# Python + SQLite3: Database Basics

In this notebook, we will use Python's built-in `sqlite3` library to:

1. Connect to a local SQLite database.
2. Create a table.
3. Insert records.
4. Retrieve records using `SELECT` and `WHERE`.
5. Update records using `UPDATE`.

**Note:** SQLite is lightweight. It stores the database in a local file, so you do not need to install or start a separate database server.

## 1. Import the library

```python
import sqlite3

print("SQLite3 imported successfully!")
```

## 2. Connect to a database

`sqlite3.connect()` connects to a database file. If the file does not exist, SQLite creates it automatically.

Here, the database file will be named `college.db` in the current working directory.

```python
connection = sqlite3.connect("college.db")

print("Database connected successfully!")
```

## 3. Create a cursor

A cursor is used to execute SQL statements and retrieve their results.

```python
cursor = connection.cursor()

print("Cursor created!")
```

## 4. Create a table

We will create a table named `students` with four columns:

- `student_id`: unique ID for each student
- `name`: student's name
- `age`: student's age
- `course`: course name

`IF NOT EXISTS` prevents an error if the table already exists.

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    course TEXT
);
"""

cursor.execute(create_table_query)
connection.commit()

print("Students table created successfully!")
```

## 5. Insert data into the table

Use `?` placeholders instead of joining values directly into the SQL string. This is safer and helps avoid SQL injection.

`executemany()` inserts multiple records using one SQL statement.

```python
students_data = [
    ("Aarav", 21, "Python"),
    ("Priya", 22, "Data Science"),
    ("Rahul", 20, "Python"),
    ("Simran", 23, "Machine Learning"),
    ("Neha", 21, "Data Analytics")
]

insert_query = """
INSERT INTO students (name, age, course)
VALUES (?, ?, ?);
"""

cursor.executemany(insert_query, students_data)
connection.commit()

print("Records inserted successfully!")
```

## 6. Retrieve all records

`SELECT *` retrieves all columns and rows from the table.

```python
cursor.execute("SELECT * FROM students")

records = cursor.fetchall()

for record in records:
    print(record)
```

## 7. Retrieve records using a WHERE clause

### Example 1: Find students enrolled in Python

The `WHERE` clause filters rows based on a condition.

```python
cursor.execute(
    "SELECT * FROM students WHERE course = ?",
    ("Python",)
)

python_students = cursor.fetchall()

for student in python_students:
    print(student)
```

### Example 2: Find students whose age is greater than 21

You can use comparison operators such as `>`, `<`, `>=`, `<=`, and `!=` in a `WHERE` clause.

```python
cursor.execute(
    "SELECT name, age, course FROM students WHERE age > ?",
    (21,)
)

older_students = cursor.fetchall()

for student in older_students:
    print(student)
```

## 8. Update records

The `UPDATE` statement modifies existing records. Always use a `WHERE` clause when you intend to update specific rows.

### Example 1: Update a student's course

Change Aarav's course from Python to Data Engineering.

```python
update_query_1 = """
UPDATE students
SET course = ?
WHERE name = ?;
"""

cursor.execute(update_query_1, ("Data Engineering", "Aarav"))
connection.commit()

print("Aarav's course updated!")
```

### Example 2: Update a student's age

Increase Rahul's age to 21.

```python
update_query_2 = """
UPDATE students
SET age = ?
WHERE name = ?;
"""

cursor.execute(update_query_2, (21, "Rahul"))
connection.commit()

print("Rahul's age updated!")
```

## 9. Verify the updated records

Run a `SELECT` query again to see the changes.

```python
cursor.execute("SELECT * FROM students")

updated_records = cursor.fetchall()

for record in updated_records:
    print(record)
```

## 10. Close the connection

Close the database connection after completing your work.

```python
cursor.close()
connection.close()

print("Database connection closed!")
```

## Quick SQL reference

| Operation | SQL statement |
|---|---|
| Create a table | `CREATE TABLE` |
| Insert records | `INSERT INTO` |
| Retrieve records | `SELECT` |
| Filter records | `WHERE` |
| Update records | `UPDATE ... SET` |
